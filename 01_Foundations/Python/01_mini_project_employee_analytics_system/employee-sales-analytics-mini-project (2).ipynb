```python
# project 01
employees = {
    101: {
        "name": "Rahul",
        "region": "North",
        "department": "Electronics",
        "sales": 85000,
        "target": 80000,
        "rating": 4.3
    },

    102: {
        "name": "Arbaz",
        "region": "West",
        "department": "Electronics",
        "sales": 125000,
        "target": 100000,
        "rating": 4.9
    },

    103: {
        "name": "John",
        "region": "South",
        "department": "Furniture",
        "sales": 65000,
        "target": 70000,
        "rating": 3.9
    },

    104: {
        "name": "Priya",
        "region": "East",
        "department": "Office Supplies",
        "sales": 98000,
        "target": 90000,
        "rating": 4.5
    }
}
#print task 01 
for keys,detail in employees.items():
    print("="*30)
    print(f"Employee id   |   {keys:}")
    print(f"Employee Name |   {detail['name']}")
    print(f"Region        |   {detail['region']}")
    print(f"Department    |   {detail['department']:}")
    print(f"Sales         |   {detail['sales']}")
    print(f"Target        |   {detail['target']}")
    print(f"Rating        |   {detail['rating']}")
    print("="*30)
# print task 02 calculate total sales
def total_sales(employees): 
    total=0
    for emp_id,employee in employees.items():
        total+=employee['sales']
    return total
result=total_sales(employees)
print("-"*39)
print(f" Total Company Sales Revenue : ${result}")
print("-"*39)

# print task 03 calculate the average each employees
def average_sales(employees):
        average=total_sales(employees)/len(employees)
        return average
avg_result=average_sales(employees)
print(f"Average sales of each Employee : ${avg_result}")

#print task 04 find highest sales employee
def highest_sales(employees):
    high_sale=0
    emp_id=''
    emp_name='' 
    for keys,detail in employees.items():
        if  detail['sales']>high_sale:
            high_sale=detail['sales']
            emp_id=keys
            emp_name=detail['name']
    return print(f"\n Highest sales :${high_sale} | Employee_id :{emp_id} | Employee Name:{emp_name} ")       
highest_sales(employees)
# print 05 lowest employee sales.
def lowest_sales(employees):
    lowest=employees[101]['sales']
    emp_name=''
    emp_id=''
    for keys,details in employees.items():
        if details['sales']<lowest:
            lowest=details['sales']
            emp_id=keys
            emp_name=details['name']
    return print(f" lowest sales :{lowest} | Employee_id :{emp_id} | Emmployee name :{emp_name}")
print(lowest_sales(employees))
 
#print 06 find out of the average above employees with using function function  
def above_average(employees):
    print("="*30)
    print(f"Above averages sales:")
    print("="*30)
    total=sum(detail['sales'] for detail in employees.values())
    average=total/len(employees)
    above_average_list=[]
    for emp_id,details in employees.items():
        if details['sales'] > average :
            above_average_list.append(f" Employee id :{emp_id} | Employees Name : {details['name' ]} |Employee Sales : ${details['sales']:.2f}")
    return "\n".join(above_average_list)
    for i in above_average_list:
        print(i)
print(above_average(employees))
# print 07 diffrent department employee sales
def department_sales(employees):
    department_data={}
    for details in employees.values():
        dept=details['department']
        sales=details['sales']
        if dept not in department_data:
            department_data[dept]={'total_sales':0,'count':0}
        department_data[dept]['total_sales']+=sales
        department_data[dept]['count']+=1

    print("=" * 40)
    print("Department Sales Analytics")
    print("=" * 40)

    for dept, data in department_data.items():
        average = data["total_sales"] / data["count"]
        print(f"Department    : {dept}")
        print(f"Total Sales   : ${data['total_sales']}")
        print(f"Employees     : {data['count']}")
        print(f"Average Sales : ${average:.2f}")
        print("-" * 40)
department_sales(employees)
#print 08 top 3 highest Sales employees
def top_employee(employees):
    print("="*35)
    print("Top 3 Highest Salary Employee")
    print("="*35)
    sorted_employees = sorted(employees.items(),key=lambda x: x[1]['sales'],reverse=True)
    result = [] 
    for emp_id, details in sorted_employees[:3]:
        result.append(f"ID: {emp_id} | Name: {details['name']} | Sales: ${details['sales']}")
    return "\n".join(result) 
print(top_employee(employees))   
 #above average sales
avg_sales = average_sales(employees)
above_avg_count = sum(1 for emp in employees.values() if emp['sales'] > avg_sales)

#print 09 of employees performance. 
def performance_employee(employees):
    print("="*30)
    print("Employees Sales report ")
    print("="*30)
    print(f"Total sales :{total_sales(employees)}")
    print(f"Average Sales:{average_sales(employees)}")
    highest_sales(employees)
    lowest_sales(employees)
    #function can store in a  varriables 
    avg_sales = average_sales(employees)
    # another varriable can create for average sales can greater then sales
    above_avg_count = sum(1 for emp in employees.values() if emp['sales'] > avg_sales)
    print(f" Above Average employees: {above_avg_count}")
performance_employee(employees)
#print 10 Employee search system
def employee_search(employees):
    print("Search For Employees")
    print("1.Search by ID")
    print("2.Search by Name")
    choice = int(input("Choose one 1 or 2 :"))
    if choice == 1:
        search_id = int(input("Enter Employee ID : "))
        for emp_id , detail in employees.items():
            if search_id == emp_id:
                return display_employee(emp_id,detail)        
    elif choice == 2:
        search_name = input("Enter Employee Name : ").lower().strip()
        for emp_id , detail in employees.items():
            if search_name == detail['name'].lower().strip():
                return display_employee(emp_id,detail)        
    return "Employee not found!"
    
# create the define the emplyees detail 
def display_employee(emp_id , detail):
    lines = [
           '='*30 ,
           f"Employee ID : {emp_id}",
           f"Name        : {detail['name']}",
           f"Sales       : {detail['sales']}",
           f"Department  : {detail['department']}",
           f"Region      : {detail['region']}",
           f"Rating      : {detail['rating']}",
           '='*30
         ]
    return "\n".join(lines)
print(employee_search(employees))
     
#print 11 employees performing rating system
def rating_system(employees):
    print('='*30)
    print("Employee Rating Performance System ")
    for emp_id,detail in employees.items():
        if detail['sales']>70000:
            detail['rating']='Excellent'
            print(display_employee(emp_id,detail))
        elif detail['sales']>50000 :
            detail['rating']='Good'
            print(display_employee(emp_id,detail))
        elif detail['sales']>30000:
            detail['rating']='Average'
            print(display_employee(emp_id,detail))
        else:
            detail['rating'] = 'Need improvement'
            print(display_employee(emp_id,detail))
rating_system(employees)
    
#print 12 Employee sales analytics system 
def analytics_system(employees):
    while True:
        menu_driven_list=["1.Total Sales","2.Highest Profermer","3.Lowest Performer","4.Above Average Employees","5.Department Analytics","6.Top 3 Employees","7.Search Employees","8.Employee Rating","9.Complete Sales Report","0.Exit"]
        print('='*30)
        print("Employee Sales Analytics System")
        print('='*30)
        for i in menu_driven_list:
            print(i)
        menu = int(input("\nEnter your choice: "))
        if menu==1:
            print(f"Total Sales :{total_sales(employees)}")
            input("Press Enter For Back To Menu")
        elif menu==2:
            print(highest_sales(employees))
            input("Press Enter For Back To Menu")
        elif menu==3:
            lowest_sales(employees)
            input("Press Enter For Back To Menu")
        elif menu==4:
            print(above_average(employees))
            input("Press Enter For Back To Menu")
        elif menu==5:
            department_sales(employees)
            input("Press Enter For Back To Menu")
        elif menu==6:
            print(top_employee(employees))
            input("Press Enter For Back To Menu")
        elif menu==7:
            search_employees(employees)
            input("Press Enter For Back To Menu")
        elif menu==8:
            rating_system(employees)
            
            input("Press Enter For Back To Menu")
        elif menu==9:
            print(performance_employee(employees))
            input("Press Enter For Back To Menu")
        elif menu==0:
            print("Thnak see For my data analytics performance")
            break
        else:
            print('Invalid choice')
analytics_system(employees)          
            
            
```
